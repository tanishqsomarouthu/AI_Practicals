```python
transitions = {
    'A' : {'Right':'B','Down':'C'},
    'B' : {'Left':'B','Down':'G'},
    'C' : {'Up':'B','Right':'G'},
    'G' : {}

}
rewards = {
    'A': -1,
    'B': -1,
    'C': -1,
    'G': 10

}


gamma=0.9
```

```python
values = {
     'A': 0,
     'B': 0,
     'C': 0,
     'G': 0


}
print(values)
```

```python
for i in range(5):
    new_values = values.copy()

    for state in ['A', 'B', 'C']:
        possible_values = []

        for action, next_state in transitions[state].items():
            value = rewards[next_state] + gamma * values[next_state]
            possible_values.append(value)

        new_values[state] = max(possible_values)

    values = new_values

print("Final values:")
print(values)
```

```python
policy = {}

for state in ['A', 'B', 'C']:
    best_action = None
    best_value = -100

    for action, next_state in transitions[state].items():
        value = rewards[next_state] + gamma * values[next_state]

        if value > best_value:
            best_value = value
            best_action = action

    policy[state] = best_action

print("Best policy:")
print(policy)
```
