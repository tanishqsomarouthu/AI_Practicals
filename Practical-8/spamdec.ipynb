```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
```

```python
messages = [
    "win a free prize now",
    "claim your free reward",
    "congratulations you won cash",
    "get a special offer today",
    "please attend the meeting tomorrow",
    "can we meet after class",
    "Your assignment is due friday",
    "Please send me the notes"
]

labels = [
    "spam",
    "spam",
    "spam",
    "spam",
    "not spam",
    "not spam",
    "not spam",
    "not spam"
]
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(messages)
```

```python
print("Vocabulary",vectorizer.vocabulary)
print(X.toarray())
```

```python
model = MultinomialNB()
model.fit(X,labels)
```

```python

new_message = ["Win cash today"]

# Convert text into numerical features
new_X = vectorizer.transform(new_message)

# Predict the message category
prediction = model.predict(new_X)

print("Message:", new_message[0])
print("Prediction:", prediction[0])
```

```python

new_message = ["claim your free reward"]
new_X = vectorizer.transform(new_message)
prediction = model.predict(new_X)
print("Message:", new_message[0])
print("Prediction:", prediction[0])
```

```python
new_message = input("Enter your message :")

new_X = vectorizer.transform([new_message])
prediction = model.predict(new_X)

print("Prediction : ",prediction[0])
```
